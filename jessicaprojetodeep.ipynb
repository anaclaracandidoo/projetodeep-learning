```python
import ollama

def simplificar_texto(texto_original):
    prompt = f"""
    Você é um assistente pedagógico especializado em acessibilidade educacional[cite: 1, 2].
    Reescreva o texto abaixo em uma linguagem simplificada e direta para uma aluna do 8º ano[cite: 1, 2].
    Use frases curtas na ordem direta[cite: 1].

    Texto Original:
    {texto_original}

    Texto Simplificado:
    """

    resposta = ollama.chat(
        model='qwen2.5:3b',
        messages=[{'role': 'user', 'content': prompt}]
    )

    return resposta['message']['content']

texto = "A Revolução Industrial acarretou em profundas transformações socioeconômicas, alterando os modos de produção."
print(simplificar_texto(texto))
```
