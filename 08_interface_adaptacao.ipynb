```python
import gradio as gr
from langchain_community.llms import Ollama
from langchain_core.prompts import ChatPromptTemplate

# 1. Inicialização da LLM
llm = Ollama(model="qwen2.5:3b", temperature=0.2)


# 2. Função de adaptação ligada à interface
def processar_adaptacao(texto_original, tipo_adaptacao):
  if not texto_original.strip():
    return "Por favor, insira um texto para ser adaptado."

  template = """
    Você é um assistente pedagógico especializado em acessibilidade educacional.
    Sua tarefa é adaptar o texto abaixo para atender às necessidades de alunos com dificuldades de aprendizagem,
    seguindo a solicitação: '{tipo_adaptacao}'.

    Texto Original:
    {texto_original}

    Texto Adaptado:
    """
  prompt = ChatPromptTemplate.from_template(template)
  chain = prompt | llm

  try:
    resultado = chain.invoke({
        "tipo_adaptacao": tipo_adaptacao,
        "texto_original": texto_original,
    })
    return resultado.strip()
  except Exception as e:
    return f"Erro ao processar adaptação: {str(e)}"


# 3. Construção da Interface Gráfica com Gradio
with gr.Blocks(theme=gr.themes.Soft(), title="EduAcessível - Adaptação") as demo:
  gr.Markdown("""
    # 📚 EduAcessível - Interface de Adaptação de Materiais
    Prototipagem do frontend para criação e adaptação de conteúdos pedagógicos acessíveis.
    """)

  with gr.Row():
    with gr.Column():
      input_texto = gr.Textbox(
          label="Texto Pedagógico Original",
          placeholder="Cole aqui o texto da aula ou exercício...",
          lines=8,
      )

      select_tipo = gr.Radio(
          choices=[
              "Linguagem Clara e Simplificada",
              "Resumo em Tópicos Curtos",
              "Adaptação para Leitura Fácil (TDHA/Autismo)",
          ],
          value="Linguagem Clara e Simplificada",
          label="Tipo de Adaptação Pretendido",
      )

      btn_adaptar = gr.Button("⚡ Gerar Material Adaptado", variant="primary")

    with gr.Column():
      output_texto = gr.Textbox(
          label="Material Adaptado (Pronto para Uso)", lines=12, interactive=False
      )

  # Ação do botão
  btn_adaptar.click(
      fn=processar_adaptacao,
      inputs=[input_texto, select_tipo],
      outputs=output_texto,
  )

# 4. Lançamento da interface (com link público temporário no Colab)
demo.launch(share=True, debug=True)
```
