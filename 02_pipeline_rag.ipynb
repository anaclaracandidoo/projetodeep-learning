```python
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. Texto de exemplo para simular o material pedagógico
texto_exemplo = """
O sistema Solar é constituído pelo Sol e por todos os corpos celestes que estão sob seu domínio gravitacional.
Ele é composto por oito planetas principais: Mercúrio, Vênus, Terra, Marte, Júpiter, Saturno, Urano e Netuno.
A Terra é o único planeta conhecido até o momento que possui vida e água em estado líquido em abundância na sua superfície.
"""

# 2. Divisão do texto em fragmentos (Chunking)
splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=20)
docs = splitter.create_documents([texto_exemplo])

# 3. Inicialização dos Embeddings via Ollama
embeddings = OllamaEmbeddings(model="nomic-embed-text")

# 4. Criar o Vector DB em memória usando ChromaDB
vector_db = Chroma.from_documents(documents=docs, embedding=embeddings)

# 5. Teste de busca por similaridade
query = "Quais são as características da Terra?"
resultados = vector_db.similarity_search(query, k=1)

print("--- Resultado da busca vetorial ---")
print(resultados[0].page_content)
```
